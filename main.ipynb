```python
import os
import pandas as pd
import logging
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from pathlib import Path
```

```python
BASE_DIR = os.path.dirname(os.path.abspath(__file__))


PASTA_DADOS = os.path.join(BASE_DIR, "dados")
PASTA_WEB = os.path.join(BASE_DIR, "web")
PASTA_RELATORIOS = os.path.join(BASE_DIR, "relatorios")
PASTA_PROCESSADOS = os.path.join(BASE_DIR, "processados")
PASTA_LOGS = os.path.join(BASE_DIR, "logs")
ARQUIVO_HTML = os.path.join(PASTA_WEB,"index.html")
URL_HTML = Path(ARQUIVO_HTML).as_uri()
ARQUIVO_ALUNOS = os.path.join(PASTA_DADOS,"alunos_projeto_final.xlsx")

print("Pasta principal:", BASE_DIR)
print("Planilha:", ARQUIVO_ALUNOS)
```

```python
os.makedirs(PASTA_LOGS, exist_ok=True)
ARQUIVO_LOG = os.path.join(PASTA_LOGS, "automacao.log")

```

```python
logging.basicConfig(
    filename=ARQUIVO_LOG,
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    encoding="utf-8"
)


logging.info("Robô iniciado com sucesso.")
logging.warning("Teste: aluno sem WhatsApp.")
logging.error("Teste: falha simulada no cadastro.")

df = pd.read_excel(
    ARQUIVO_ALUNOS,
    dtype={
        "CPF": str,
        "WhatsApp": str
    }
)
```

```python
# Verifica se o arquivo HTML existe
if os.path.exists(ARQUIVO_HTML):

    logging.info("Arquivo HTML localizado.")

    # Inicializa o Google Chrome
    driver = webdriver.Chrome()

    try:
        # Abre a página HTML local
        driver.get(URL_HTML)

        logging.info("Página HTML aberta no Chrome.")

        print("Título da página:", driver.title)
        print("URL:", driver.current_url)

        # Aguarda a presença do campo Nome
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.ID, "nome"))
        )

        print("Campo Nome encontrado com sucesso!")

    except Exception:
        logging.exception("Erro ao abrir ou analisar a página HTML.")
        raise

    finally:
        # Encerra o navegador mesmo se ocorrer um erro
        driver.quit()

else:
    logging.error("O arquivo index.html não foi encontrado.")
    print("ERRO: Arquivo HTML não encontrado.")
```
